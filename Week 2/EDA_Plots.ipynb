```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud
import squarify

sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Temperature_Anomaly"],
        bins=40,
            kde=True
            )

plt.title("Distribution of Temperature Anomalies")
plt.xlabel("Temperature Anomaly (°C)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
        x="Temperature_Anomaly",
            fill=True
            )
plt.title("Density of Global Temperature Anomalies")
plt.xlabel("Temperature Anomaly (°C)")
plt.ylabel("Density")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    y=df["Temperature_Anomaly"]
    )
plt.title("Spread and Outliers of Temperature Anomalies")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    y=df["Temperature_Anomaly"]
    )
plt.title("Shape of Temperature Anomaly Distribution")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df.sample(min(5000, len(df)), random_state=42),
        x="CO2_Emissions",
            y="Temperature_Anomaly",
                alpha=0.5
                )

plt.title("CO₂ Emissions vs Temperature Anomaly")
plt.xlabel("CO₂ Emissions")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df.sample(min(3000, len(df)), random_state=42),
        x="CO2_Emissions",
            y="Temperature_Anomaly",
                scatter_kws={"alpha": 0.3},
                    line_kws={"linewidth": 3}
                    )

plt.title("Regression Between CO₂ Emissions and Temperature Anomaly")
plt.xlabel("CO₂ Emissions")
plt.ylabel("Temperature Anomaly (°C)")

plt.show()
```

```python
country_anomaly = (
      df.groupby("Country")["Temperature_Anomaly"]
          .mean()
              .dropna()
              )

              # Remove negative values for word-cloud weighting
weights = country_anomaly - country_anomaly.min() + 0.01

wordcloud = WordCloud(
                  width=1200,
                      height=600,
                          background_color="white",
                              colormap="plasma"
                              ).generate_from_frequencies(weights.to_dict())
plt.figure(figsize=(15, 8))

plt.imshow(
                                  wordcloud,
                                      interpolation="bilinear"
                                      )

plt.axis("off")

plt.title(
                                          "Countries with Higher Average Temperature Anomalies",
                                              fontsize=18
                                              )

plt.show()
```

```python
top_countries = (
      df.groupby("Country")["Temperature_Anomaly"]
          .mean()
              .sort_values(ascending=False)
                  .head(15)
                  )

plt.figure(figsize=(12, 8))

sns.barplot(
                      x=top_countries.values,
                          y=top_countries.index
                          )

plt.title("Top 15 Countries by Average Temperature Anomaly")
plt.xlabel("Average Temperature Anomaly (°C)")
plt.ylabel("Country")
plt.show()
```

```python
country_data = (
      df.groupby("Country")["Temperature_Anomaly"]
          .mean()
              .abs()
                  .nlargest(20)
                  )
plt.figure(figsize=(14, 9))

squarify.plot(
                      sizes=country_data.values,
                          label=country_data.index,
                              alpha=0.8
                              )

plt.title("Top 20 Countries by Temperature Anomaly Magnitude")
plt.axis("off")

plt.show()
```

```python
yearly = (
      df.groupby("Year")["Temperature_Anomaly"]
          .mean()
          )
plt.figure(figsize=(14, 6))

plt.plot(
              yearly.index,
                  yearly.values,
                      marker="o"
                      )

plt.title("Global Temperature Anomaly Over the Years")
plt.xlabel("Year")
plt.ylabel("Temperature Anomaly (°C)")
plt.show()
```

```python
df["Decade"] = (df["Year"] // 10) * 10

decade = (
    df.groupby("Decade")["Temperature_Anomaly"]
        .mean()
        )

plt.figure(figsize=(12, 6))

plt.bar(
            decade.index.astype(str),
                decade.values
                )

plt.title("Average Temperature Anomaly by Decade")
plt.xlabel("Decade")
plt.ylabel("Temperature Anomaly (°C)")

plt.show()
```

```python
top15 = (
      df.groupby("Country")["Temperature_Anomaly"]
          .mean()
              .nlargest(15)
                  .index
                  )

heat_data = df[
                      df["Country"].isin(top15)
                      ].pivot_table(
                          values="Temperature_Anomaly",
                              index="Country",
                                  columns="Year",
                                      aggfunc="mean"
                                      )

                                      plt.figure(figsize=(16, 9))

                                      sns.heatmap(
                                          heat_data,
                                              cmap="RdYlBu_r",
                                                  center=0
                                                  )

                                                  plt.title("Temperature Anomaly Across Top 15 Countries Over Time")
                                                  plt.xlabel("Year")
                                                  plt.ylabel("Country")

                                                  plt.show()
)
```
